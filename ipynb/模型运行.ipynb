# XGBoost 模型批量运行

这个notebook用于批量运行XGBoost文件夹中的所有模型文件

```python
import os
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import warnings
import json

# import import_ipynb

warnings.filterwarnings('ignore')

def run_notebook(notebook_path, params=None):
    """运行指定的notebook文件"""
    print(f"正在运行: {notebook_path}")
    
    try:
        # 读取notebook
        with open(notebook_path, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
        
        if params:
            param_code = "\n".join([f"{k} = {repr(v)}" for k, v in params.items()])
            param_cell = nbformat.v4.new_code_cell(param_code)
            print(param_code)
            nb.cells.insert(0, param_cell)

        # 创建执行器
        ep = ExecutePreprocessor(timeout=1800, kernel_name='python3')
        
        # 执行notebook
        ep.preprocess(nb, {'metadata': {'path': os.path.dirname(notebook_path)}})
        
        print(f"✓ 成功完成: {os.path.basename(notebook_path)}")
        return True
        
    except Exception as e:
        print(f"✗ 运行失败: {os.path.basename(notebook_path)} - {str(e)}")
        return False
```

```python
folder_path = '/Users/yanjieni/Desktop/projects/脑脊液/naojiye/data/final_data2' 
all_items = os.listdir(folder_path)
datasets = [f for f in all_items if os.path.isfile(os.path.join(folder_path, f))]
# datasets = [
#  'filtered_patient2.csv',]
#  'filtered_patient7.csv']
datasets
```

```python
def run_models(file_path):
    # 设置XGBoost文件夹路径
    # notebook_files1 = list_all_files("/Users/yanjieni/Desktop/脑脊液/naojiye/ipynb/XGBoost")
    def list_all_files(root_dir):
        all_files = []
        for root, dirs, files in os.walk(root_dir):
            for file in files:
                if file.endswith(".ipynb"):
                    all_files.append(os.path.join(root, file))
        return all_files

    notebook_files1 = list_all_files(file_path)
    notebook_files1 = [file_path]
    for i in notebook_files1:
        if ".DS_Store" in i:
            notebook_files1.remove(i)
            break
        
    print(f"找到 {len(notebook_files1)} 个notebook文件:")

    total_count = len(notebook_files1)*len(datasets)

    for i, file in enumerate(notebook_files1, 1):
        print(f"{i}. {file}")
        
    success_count = 0

    for dataset in datasets:

        for notebook_file in notebook_files1:
            # notebook_path = os.path.join(xgboost_folder1, notebook_file)
            
            if run_notebook(notebook_file, params={"dataset": dataset}):
                success_count += 1
            
            print("-" * 50)

    print(f"\n运行总结:")
    print(f"总文件数: {total_count}")
    print(f"成功运行: {success_count}")
    print(f"失败数量: {total_count - success_count}")
```
