# XGBoost 模型批量运行

这个notebook用于批量运行XGBoost文件夹中的所有模型文件

```python
import os
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import warnings
import json

# import import_ipynb

warnings.filterwarnings('ignore')

def run_notebook(notebook_path, params=None):
    """运行指定的notebook文件"""
    print(f"正在运行: {notebook_path}")
    
    try:
        # 读取notebook
        with open(notebook_path, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
        
        if params:
            param_code = "\n".join([f"{k} = {repr(v)}" for k, v in params.items()])
            param_cell = nbformat.v4.new_code_cell(param_code)
            print(param_code)
            nb.cells.insert(0, param_cell)

        # 创建执行器
        ep = ExecutePreprocessor(timeout=1800, kernel_name='python3')
        
        # 执行notebook
        ep.preprocess(nb, {'metadata': {'path': os.path.dirname(notebook_path)}})
        
        print(f"✓ 成功完成: {os.path.basename(notebook_path)}")
        return True
        
    except Exception as e:
        print(f"✗ 运行失败: {os.path.basename(notebook_path)} - {str(e)}")
        return False
```

```python
def run_models(file_path):
    # 设置XGBoost文件夹路径
    # notebook_files1 = list_all_files("/Users/yanjieni/Desktop/脑脊液/naojiye/ipynb/XGBoost")
    def list_all_files(root_dir):
        all_files = []
        for root, dirs, files in os.walk(root_dir):
            for file in files:
                if file.endswith(".ipynb"):
                    all_files.append(os.path.join(root, file))
        return all_files

    notebook_files1 = list_all_files(file_path)
    notebook_files1 = [file_path]
    for i in notebook_files1:
        if ".DS_Store" in i:
            notebook_files1.remove(i)
            break
        
    print(f"找到 {len(notebook_files1)} 个notebook文件:")

    total_count = len(notebook_files1)*len(datasets)

    for i, file in enumerate(notebook_files1, 1):
        print(f"{i}. {file}")
        
    success_count = 0

    for dataset in datasets:

        for notebook_file in notebook_files1:
            # notebook_path = os.path.join(xgboost_folder1, notebook_file)
            
            if run_notebook(notebook_file, params={"dataset": dataset}):
                success_count += 1
            
            print("-" * 50)

    print(f"\n运行总结:")
    print(f"总文件数: {total_count}")
    print(f"成功运行: {success_count}")
    print(f"失败数量: {total_count - success_count}")
```

```python
import pandas as pd
folder = "/Users/yanjieni/Desktop/脑脊液/naojiye/data/model_data2"
all_files = []
for root, dirs, files in os.walk(folder):
    for file in files:
        all_files.append(os.path.join(root, file))

# 数据输出
output_file = "/Users/yanjieni/Desktop/脑脊液/naojiye/模型训练结果3.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for file in all_files:
        data = pd.read_json(file)
        records = []

        for dataset_name, comparisons in data.items():
            for comparison_name, metrics in comparisons.items():
                row = {"数据集": dataset_name, "任务组": comparison_name}
                row.update(metrics)
                records.append(row)

        df = pd.DataFrame(records)

        df.sort_values(['数据集'])
        comparison_order = [
            "阴性组vs其他",
            "感染性脑膜炎v非感染组",
            "细菌性脑膜炎vs其他感染性脑膜炎",
            "病毒性脑膜炎vs其他感染性脑膜炎",
            "结核性脑膜炎vs其他感染性脑膜炎",
            "真菌性脑膜炎vs其他感染性脑膜炎",
            "肿瘤性脑膜炎vs其他",
            "自身免疫性脑膜炎vs其他",
            "病毒性脑膜炎vs病毒性脑炎"
        ]

        df["任务组"] = pd.Categorical(df["任务组"], categories=comparison_order, ordered=True)

        df = df.sort_values(by=["数据集", "任务组"]).reset_index(drop=True)
        
        sheet = ((file.split("/")[-1]).split(".")[0]).split("_")[0]
        sheet_name = f"{sheet}模型结果"  
        df.to_excel(writer, sheet_name=sheet_name, index=False)

        print(f"已保存到 {output_file} 的 sheet: {sheet_name}")
```
